```python
import numpy as np
import time

prices = np.array([100, 102, 101, 105, 107, 106, 110, 108, 112, 115])

start = time.time()
loop_returns = []
for i in range(1, len(prices)):
    pct_change = (prices[i] - prices[i-1]) / prices[i-1]
    loop_returns.append(pct_change)
loop_time = time.time() - start

start = time.time()
vectorized_returns = (prices[1:] - prices[:-1]) / prices[:-1]
vectorized_time = time.time() - start

print("Loop result:", loop_returns)
print("Vectorized result:", vectorized_returns)
print(f"Loop time: {loop_time:.8f}s | Vectorized time: {vectorized_time:.8f}s")
```
